```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage
import os
```

```python
os.makedirs("../models", exist_ok=True)
os.makedirs("../reports/figures", exist_ok=True)

df = pd.read_csv("../data/processed/subdivision_pattern.csv", index_col=0)
X = StandardScaler().fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)
result = pd.DataFrame({"SUBDIVISION": df.index, "CLUSTER": labels})
print(result)
result.to_csv("../models/obj3_kmeans_clusters.csv", index=False)
```

```python
plt.figure(figsize=(10,8))
for cluster, data in df.groupby(labels):
    plt.plot(data.mean(), marker="o", label=f"Cluster {cluster}")

plt.title("Average Monthly Rainfall Pattern")
plt.xlabel("Month")
plt.ylabel("Rainfall (mm)")
plt.legend()
plt.grid()
plt.savefig("../reports/figures/obj3_kmeans_pattern.png")
plt.close()
```

```python
Z = linkage(X, method="ward")
dendrogram(Z, labels=df.index, leaf_rotation=90, leaf_font_size=7)

plt.title("Hierarchical Clustering Dendrogram")
plt.tight_layout()
plt.savefig("../reports/figures/obj3_dendrogram.png")
plt.close()
```
