```python
import pandas as pd
import os
```

```python
RAW_PATH = "C:/Users/HP/Desktop/Predictive_Analysis/data/sub-division_rainfall_act_dep_1997-2015 (2).csv"
CLEAN_PATH = "C:/Users/HP/Desktop/Predictive_Analysis/data/cleaned_rainfall.csv"
```

```python
df = pd.read_csv(RAW_PATH)
print(df.head())
print("Original shape:", df.shape)
```

```python
# Keep only actual rainfall records
df = df[df["Parameter"] == "Actual"].copy()
df.drop(columns=["Parameter"], inplace=True)
print("After filtering:", df.shape)
```

```python
# Clean YEAR column
df["YEAR"] = pd.to_numeric(df["YEAR"], errors="coerce")
df = df.dropna(subset=["YEAR"])
df["YEAR"] = df["YEAR"].astype(int)
```

```python
# Fill missing numerical values with column medians
columns = [
    "JAN", "FEB", "MAR", "APR", "MAY", "JUN",
    "JUL", "AUG", "SEP", "OCT", "NOV", "DEC",
    "JF", "MAM", "JJAS", "OND", "ANNUAL"
]

df[columns] = df[columns].fillna(df[columns].median())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Classify annual rainfall
low = df["ANNUAL"].quantile(0.25)
high = df["ANNUAL"].quantile(0.75)

df["RAINFALL_CATEGORY"] = "Normal"
df.loc[df["ANNUAL"] < low, "RAINFALL_CATEGORY"] = "Deficient"
df.loc[df["ANNUAL"] > high, "RAINFALL_CATEGORY"] = "Excess"
```

```python
# Save cleaned dataset
os.makedirs(os.path.dirname(CLEAN_PATH), exist_ok=True)
df.to_csv(CLEAN_PATH, index=False)

print("Cleaned shape:", df.shape)
print("\nRainfall categories:")
print(df["RAINFALL_CATEGORY"].value_counts())

print("\nFirst five rows:")
print(df.head())
```
