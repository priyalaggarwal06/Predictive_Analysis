####Feature Engineering

```python
import pandas as pd
import os
```

```python
# Load cleaned data
df = pd.read_csv("C:/Users/HP/Desktop/Predictive_Analysis/data/cleaned_rainfall.csv")
```

```python
# Create rainfall tables
jjas = df.pivot(index="YEAR", columns="SUBDIVISION", values="JJAS")
annual = df.pivot(index="YEAR", columns="SUBDIVISION", values="ANNUAL")
```

```python
# Prepare Punjab dataset
neighbours = ["HARYANA DELHI & CHANDIGARH","HIMACHAL PRADESH","JAMMU & KASHMIR","WEST RAJASTHAN","EAST RAJASTHAN","WEST UTTAR PRADESH","UTTARAKHAND"]
punjab = pd.DataFrame({
    "YEAR": jjas.index,
    "PUNJAB_JJAS": jjas["PUNJAB"],
    "PUNJAB_ANNUAL": annual["PUNJAB"]
})
for n in neighbours:
    punjab[n + "_JJAS"] = jjas[n]
punjab = punjab.dropna().copy()
```

```python
# Create drought labels
threshold = punjab["PUNJAB_JJAS"].quantile(0.25)
punjab["DROUGHT"] = (punjab["PUNJAB_JJAS"] < threshold).astype(int)
```

```python
# Save Punjab dataset
os.makedirs("data", exist_ok=True)
punjab.to_csv("data/punjab_dataset.csv", index=False)
```

```python
# Prepare data for clustering
months = ["JAN", "FEB", "MAR", "APR", "MAY", "JUN","JUL", "AUG", "SEP", "OCT", "NOV", "DEC"]
pattern = df.groupby("SUBDIVISION")[months].mean()
pattern.to_csv("data/subdivision_pattern.csv")
```
