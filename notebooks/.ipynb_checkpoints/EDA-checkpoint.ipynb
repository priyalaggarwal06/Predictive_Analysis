```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
CLEAN_PATH = "C:/Users/HP/Desktop/Predictive_Analysis/data/cleaned_rainfall.csv"
df = pd.read_csv(CLEAN_PATH)
os.makedirs("images", exist_ok=True)
print(df.describe())
```

```python
# 1. Annual distribution
plt.figure()
plt.hist(df["ANNUAL"], edgecolor="black")
plt.title("Annual Rainfall Distribution")
plt.xlabel("Annual Rainfall (mm)")
plt.ylabel("Count")
plt.show()
```

```python
# 2. Monthly average (main insight)
monthly = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN','JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
plt.figure(figsize=(9,5))
plt.bar(monthly, df[monthly].mean())
plt.title("Average Monthly Rainfall")
plt.xlabel("Month")
plt.ylabel("Average Rainfall (mm)")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(df["JJAS"], df["ANNUAL"], alpha=0.4, s=10)
plt.title("Monsoon Rainfall vs Annual Rainfall")
plt.xlabel("JJAS Rainfall (mm)")
plt.ylabel("Annual Rainfall (mm)")
plt.show()
```

```python
# 4. Yearly trend 1901-2015
yearly = df.groupby("YEAR")["ANNUAL"].mean()
plt.plot(yearly.index, yearly.values)
plt.title("Yearly Rainfall Trend")
plt.xlabel("Year")
plt.ylabel("Average Annual Rainfall (mm)")
plt.show()
```

```python
# 5. Top 10 wettest subdivisions
top = df.groupby("SUBDIVISION")["ANNUAL"].mean().nlargest(10)
plt.barh(top.index, top.values)
plt.title("Top 10 Wettest Subdivisions")
plt.xlabel("Average Annual Rainfall (mm)")
plt.show()
```
